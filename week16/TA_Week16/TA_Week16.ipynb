### **Parameter간 correlation을 보기 위한 contour 출력**

Contour 그리기에 다소 어려움을 겪고 있을 것 같아 제가 생각하는 방법을 대략적으로 작성했습니다.

---
### Method 1 : HW21에서 contour를 그린 방식
HW21 에서는 $\rho_0$, $r_s$ 두 개의 파라미터에 대해 fitting을 했습니다. 그래서 2차원 평면에서 log-likelihood(~posterior, $\chi^2$)를 계산해 ~68, 95, 99% 에 해당하는 $1\sigma$, $2\sigma$, $3\sigma$ contour를 그렸습니다.

이번 과제는 parameter가 4개이기 때문에 likelihood는 4dimension에 존재합니다. 
따라서 correlation을 보고 싶은 2D plane (e.g. $\rho_1$ - $r_{s1}$) 에서 likelihood를 계산할 수 없습니다 (나머지 $\rho_2$, $r_{s2}$ 에 대한 dimension이 남아있기 때문에). 

추천하는 방법은 Lavenberg-Marquardt method로 구한 best-fit parameter로 나머지 dimension을 projection 하는 것입니다. 
그러면 HW21에서와 동일한 방법으로 contour를 그릴 수 있습니다.

---
### Method 2 : MCMC sampling + `getdist` or `corner`

MCMC sampling은 posterior 분포를 바탕으로 샘플링하는 방법입니다. 자세한 과정에 대한 이해보다는 우리가 구한 확률 분포에 맞추어 파라미터를 샘플링한다고 생각하면 좋습니다. (학기 초에 rejection method를 통해 gaussian distribution을 따르는 random number를 샘플링한 것처럼)

따라서 1만개를 샘플링하면 (10000, 4) shape array를 얻을 수 있고 2d histogram을 그리고 contour를 그릴 수 있습니다. 

MCMC sampling은 `emcee`, contour를 그리는 library는 `getdist` 또는 `corner`를 활용할 수 있습니다.
